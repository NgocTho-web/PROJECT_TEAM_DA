```python
import pandas as pd

df_static_1 = pd.read_csv('data/raw/blackwell_shop_cleaned.csv', dtype=str)
df_static_2 = pd.read_csv('data/raw/web_crawling_tiki.csv', dtype=str)
df_dynamic = pd.read_csv('data/raw/dynamic_dataset.csv', dtype=str)

df_static_2 = df_static_2.rename(columns={
    'ID': 'isbn',
    'Tên sách': 'name',
    'Tác giả': 'author',
    'Nhà xuất bản': 'publisher',
    'Giá': 'gbpprice'
})

unified_static_master = pd.concat([df_static_1, df_static_2], ignore_index=True)

integrated_master_dataset = pd.merge(
    left=unified_static_master, 
    right=df_dynamic, 
    left_on='isbn',
    right_on='ID',
    how='left'
)

integrated_master_dataset.to_csv('data/integrated/Unified_Integrated_Master_Dataset.csv', index=False)
```
