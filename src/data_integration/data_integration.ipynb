```python

import pandas as pd

# 1. Đọc dữ liệu
df_static_1 = pd.read_csv(
    '../../data/raw/blackwell_shop_cleaned.csv', dtype=str
)
df_static_2 = pd.read_csv(
    '../../data/raw/web_crawling_tiki.csv', dtype=str
)
df_dynamic = pd.read_csv(
    '../../data/raw/dynamic_dataset.csv', dtype=str
)
df_llm = pd.read_csv(
    '../../data/raw/llm_dataset.csv', dtype=str
)

# 2. Chuẩn hóa tên cột
rename_columns = {
    'ID': 'isbn',
    'Tên sách': 'name',
    'Tên sản phẩm': 'name',
    'Tác giả': 'author',
    'Nhà xuất bản': 'publisher',
    'Giá': 'gbpprice',
    'Giá gốc': 'original_price',
    'Giảm giá': 'discount_price',
    'Phần trăm giảm': 'discount_percentage',
    'Đánh giá': 'rating',
    'Số lượt đánh giá': 'rating_count',
    'LLM Content': 'llm_content'
}

df_static_1 = df_static_1.rename(columns=rename_columns)
df_static_2 = df_static_2.rename(columns=rename_columns)
df_llm = df_llm.rename(columns=rename_columns)

# 3. Gộp 3 nguồn dữ liệu tĩnh
unified_static_master = pd.concat(
    [df_static_1, df_static_2, df_llm],
    ignore_index=True
)

# 4. Tích hợp dữ liệu động
integrated_master_dataset = pd.merge(
    left=unified_static_master,
    right=df_dynamic,
    left_on='isbn',
    right_on='ID',
    how='left'
)

# 5. Xuất dữ liệu
integrated_master_dataset.to_csv(
    '../../data/integrated/Unified_Integrated_Master_Dataset.csv',
    index=False
)

print('Static dataset shape:', unified_static_master.shape)
print('Integrated dataset shape:', integrated_master_dataset.shape)
print('Đã xuất dữ liệu thành công!')
```
