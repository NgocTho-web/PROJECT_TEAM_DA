```python
from pathlib import Path

import pandas as pd


# 1. PATH

ROOT_DIR = Path(__file__).resolve().parents[2]

DATA_PATH = ROOT_DIR / "data" / "blackwell_shop_cleaned.csv"

REPORT_PATH = (
    ROOT_DIR
    / "reports"
    / "member5_final_report.md"
)



# 2. LOAD DATA

df = pd.read_csv(DATA_PATH)

df["no_of_pages_num"] = pd.to_numeric(
    df["no_of_pages"],
    errors="coerce"
)



# 3. CALCULATE INSIGHTS

numeric_cols = [
    "gbpprice",
    "discount",
    "height",
    "width",
    "spine",
    "weight",
    "salesRank",
    "no_of_pages_num"
]

numeric_cols = [
    col for col in numeric_cols
    if col in df.columns
]

corr = df[numeric_cols].corr(
    method="pearson"
)

category_price = (
    df.groupby("category")["gbpprice"]
    .mean()
    .sort_values(ascending=False)
)

top_category = category_price.index[0]
top_category_price = category_price.iloc[0]

top_publisher = (
    df["publisher"]
    .value_counts()
    .idxmax()
)

top_type = (
    df["type"]
    .value_counts()
    .idxmax()
)



# 4. FINAL REPORT

report = f"""
# MEMBER 5 - MULTIVARIATE ANALYSIS + DATA VISUALIZATION + INSIGHTS

## 1. Multivariate Analysis

### Numerical - Numerical

Pearson và Spearman correlation được sử dụng để phân tích
mối quan hệ giữa các biến numerical.

Các biến được phân tích gồm:

- gbpprice
- discount
- height
- width
- spine
- weight
- salesRank
- no_of_pages

Heatmap được sử dụng để trực quan hóa ma trận tương quan.

### Categorical - Numerical

Phân tích giá sách theo từng category bằng:

- Mean
- Median
- Standard deviation
- Minimum
- Maximum

Kết quả cho thấy category có mức giá trung bình cao nhất
trong dữ liệu là **{top_category}**, với giá trung bình khoảng
**{top_category_price:.2f} GBP**.

### Categorical - Categorical

Cross-tabulation được sử dụng để phân tích mối quan hệ
giữa `category` và `type`.



# 2. Data Visualization

Các visualization được xây dựng gồm:

1. Pearson Correlation Heatmap
2. Scatter Plot
3. Grouped Bar Chart
4. Grouped Box Plot
5. Stacked Bar Chart
6. Histogram + KDE
7. Univariate Box Plot
8. Bar Chart
9. Pie Chart

Các biểu đồ đều được chuẩn hóa với:

- Title
- X Label
- Y Label
- Legend phù hợp



# 3. Insights - So What?

## Heatmap

Heatmap giúp xác định mức độ tương quan giữa các biến numerical.
Các biến có correlation cao cần được chú ý khi xây dựng mô hình.

So What?

Correlation giúp xác định những biến có khả năng mang thông tin
trùng lặp và hỗ trợ lựa chọn features.



## Scatter Plot

Scatter Plot thể hiện mối quan hệ giữa số trang và giá sách.

So What?

Biểu đồ giúp đánh giá liệu số trang có xu hướng liên quan
đến giá sách hay không và có thể phát hiện quan hệ phi tuyến
hoặc các điểm bất thường.



## Grouped Bar Chart

Grouped Bar Chart so sánh giá trung bình giữa các category.

So What?

Sự khác biệt về giá trung bình giữa các category cho thấy
category có thể là một feature hữu ích khi phân tích giá sách.



## Grouped Box Plot

Box Plot thể hiện phân phối giá giữa các category.

So What?

Biểu đồ giúp phát hiện sự khác biệt về median,
độ phân tán và outliers giữa các nhóm.



## Stacked Bar Chart

Stacked Bar Chart thể hiện phân bố các loại sách
trong từng category.

So What?

Cấu trúc loại sách khác nhau giữa các category
có thể cung cấp thông tin cho feature engineering.



## Histogram + KDE

Histogram và KDE thể hiện phân phối của giá sách.

So What?

Phân phối dữ liệu giúp xác định skewness,
outliers và lựa chọn phương pháp preprocessing phù hợp.



## Box Plot

Box Plot của giá sách cho phép phát hiện
các giá trị bất thường.

So What?

Outliers có thể ảnh hưởng đến một số mô hình Machine Learning,
đặc biệt là các mô hình nhạy với scale và phân phối dữ liệu.



# 4. AI / ML Model Implication

Kết quả Multivariate EDA có một số implications:

- Cần kiểm tra correlation trước khi sử dụng các biến numerical.
- Categorical variables cần được encoding.
- Outliers cần được kiểm tra trước khi huấn luyện model.
- Nếu sử dụng Linear Regression, cần chú ý multicollinearity.
- Các mô hình tree-based như Random Forest hoặc Gradient Boosting
  có thể phù hợp hơn khi dữ liệu có quan hệ phi tuyến.
- Feature engineering có thể sử dụng category, type, price,
  number of pages và các đặc điểm vật lý của sách.



# 5. Overall Insights

- Category có sự khác biệt về mức giá trung bình.
- Giá sách có sự phân tán và có thể xuất hiện outliers.
- Các biến numerical có mức độ tương quan khác nhau.
- Category và type có sự phân bố khác nhau.
- Publisher xuất hiện nhiều nhất là **{top_publisher}**.
- Book type phổ biến nhất là **{top_type}**.



# 6. Final Conclusion

Multivariate Analysis kết hợp với Data Visualization giúp
xác định mối quan hệ giữa numerical variables,
sự khác biệt giữa các categorical groups và phân phối dữ liệu.

Các insight thu được có thể được sử dụng cho bước tiếp theo:
Feature Engineering và lựa chọn Machine Learning model.
"""



# 5. SAVE REPORT


REPORT_PATH.parent.mkdir(
    parents=True,
    exist_ok=True
)

with open(
    REPORT_PATH,
    "w",
    encoding="utf-8"
) as file:
    file.write(report)


print("=" * 70)
print("INSIGHTS + FINAL REPORT COMPLETED")
print("=" * 70)

print(f"\nReport saved at:")
print(REPORT_PATH)
```
