```python
from pathlib import Path
import pandas as pd
import numpy as np


# 1. PATH

ROOT_DIR = Path(__file__).resolve().parents[2]

DATA_PATH = ROOT_DIR / "data" / "blackwell_shop_cleaned.csv"


# 2. LOAD DATA

df = pd.read_csv(DATA_PATH)

print("=" * 70)
print("MEMBER 5 - MULTIVARIATE ANALYSIS")
print("=" * 70)

print("\nDataset shape:")
print(df.shape)

print("\nColumns:")
print(df.columns.tolist())



# 3. PREPARE DATA

df["no_of_pages_num"] = pd.to_numeric(
    df["no_of_pages"],
    errors="coerce"
)



# 4. NUMERICAL - NUMERICAL

numeric_cols = [
    "gbpprice",
    "discount",
    "height",
    "width",
    "spine",
    "weight",
    "salesRank",
    "no_of_pages_num"
]

numeric_cols = [
    col for col in numeric_cols
    if col in df.columns
]

pearson_corr = df[numeric_cols].corr(
    method="pearson"
)

print("\n" + "=" * 70)
print("PEARSON CORRELATION")
print("=" * 70)

print(pearson_corr.round(2))



# 5. SPEARMAN CORRELATION

spearman_corr = df[numeric_cols].corr(
    method="spearman"
)

print("\n" + "=" * 70)
print("SPEARMAN CORRELATION")
print("=" * 70)

print(spearman_corr.round(2))



# 6. CATEGORICAL - NUMERICAL

print("\n" + "=" * 70)
print("CATEGORICAL - NUMERICAL")
print("=" * 70)

category_statistics = (
    df.groupby("category")["gbpprice"]
    .agg(
        ["count", "mean", "median", "std", "min", "max"]
    )
    .sort_values("mean", ascending=False)
)

print(category_statistics.head(10))



# 7. CATEGORICAL - CATEGORICAL
# CROSS TABULATION

print("\n" + "=" * 70)
print("CATEGORICAL - CATEGORICAL")
print("=" * 70)

top_categories = (
    df["category"]
    .value_counts()
    .head(10)
    .index
)

top_types = (
    df["type"]
    .value_counts()
    .head(5)
    .index
)

cross_df = df[
    df["category"].isin(top_categories)
    & df["type"].isin(top_types)
]

cross_table = pd.crosstab(
    cross_df["category"],
    cross_df["type"]
)

print("\nCross-tabulation:")
print(cross_table)



# 8. CONTINGENCY TABLE - PERCENTAGE

contingency_table = pd.crosstab(
    cross_df["category"],
    cross_df["type"],
    normalize="index"
) * 100

print("\nContingency Table (%):")
print(contingency_table.round(2))


print("\n" + "=" * 70)
print("MULTIVARIATE ANALYSIS COMPLETED")
print("=" * 70)
```
