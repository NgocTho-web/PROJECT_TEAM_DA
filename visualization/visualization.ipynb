```python
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. PATH

ROOT_DIR = Path(__file__).resolve().parents[2]

DATA_PATH = ROOT_DIR / "data" / "blackwell_shop_cleaned.csv"

OUTPUT_DIR = ROOT_DIR / "visualization" / "member5_outputs"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)


# 2. LOAD DATA

df = pd.read_csv(DATA_PATH)

print("=" * 70)
print("MEMBER 5 - DATA VISUALIZATION")
print("=" * 70)



# 3. PREPARE DATA

df["no_of_pages_num"] = pd.to_numeric(
    df["no_of_pages"],
    errors="coerce"
)


# CHART 1 - HEATMAP

numeric_cols = [
    "gbpprice",
    "discount",
    "height",
    "width",
    "spine",
    "weight",
    "salesRank",
    "no_of_pages_num"
]

numeric_cols = [
    col for col in numeric_cols
    if col in df.columns
]

corr = df[numeric_cols].corr(method="pearson")

plt.figure(figsize=(11, 8))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    center=0
)

plt.title("Pearson Correlation Heatmap")
plt.xlabel("Numerical Variables")
plt.ylabel("Numerical Variables")

plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "01_heatmap.png",
    dpi=300
)

plt.show()



# CHART 2 - SCATTER PLOT

scatter_df = df[
    ["no_of_pages_num", "gbpprice"]
].dropna()

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=scatter_df,
    x="no_of_pages_num",
    y="gbpprice",
    alpha=0.5
)

plt.title(
    "Number of Pages vs Book Price"
)

plt.xlabel("Number of Pages")
plt.ylabel("Price (GBP)")

plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "02_scatter_plot.png",
    dpi=300
)

plt.show()



# CHART 3 - GROUPED BAR CHART

category_price = (
    df.groupby("category")["gbpprice"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=category_price,
    x="category",
    y="gbpprice"
)

plt.title(
    "Average Book Price by Category"
)

plt.xlabel("Category")
plt.ylabel("Average Price (GBP)")

plt.xticks(
    rotation=45,
    ha="right"
)

plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "03_grouped_bar.png",
    dpi=300
)

plt.show()



# CHART 4 - GROUPED BOX PLOT

top_categories = (
    df["category"]
    .value_counts()
    .head(8)
    .index
)

box_df = df[
    df["category"].isin(top_categories)
][
    ["category", "gbpprice"]
].dropna()

plt.figure(figsize=(12, 7))

sns.boxplot(
    data=box_df,
    x="category",
    y="gbpprice"
)

plt.title(
    "Book Price Distribution by Category"
)

plt.xlabel("Category")
plt.ylabel("Price (GBP)")

plt.xticks(
    rotation=45,
    ha="right"
)

plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "04_grouped_boxplot.png",
    dpi=300
)

plt.show()



# CHART 5 - STACKED BAR CHART

top_categories = (
    df["category"]
    .value_counts()
    .head(6)
    .index
)

top_types = (
    df["type"]
    .value_counts()
    .head(5)
    .index
)

cross_df = df[
    df["category"].isin(top_categories)
    & df["type"].isin(top_types)
]

cross_table = pd.crosstab(
    cross_df["category"],
    cross_df["type"]
)

cross_table.plot(
    kind="bar",
    stacked=True,
    figsize=(12, 7)
)

plt.title(
    "Book Type Distribution Across Categories"
)

plt.xlabel("Category")
plt.ylabel("Number of Books")

plt.xticks(
    rotation=45,
    ha="right"
)

plt.legend(
    title="Book Type",
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "05_stacked_bar.png",
    dpi=300
)

plt.show()



# CHART 6 - HISTOGRAM + KDE

price_df = df["gbpprice"].dropna()

plt.figure(figsize=(10, 6))

sns.histplot(
    price_df,
    bins=30,
    kde=True
)

plt.title(
    "Distribution of Book Prices"
)

plt.xlabel("Price (GBP)")
plt.ylabel("Frequency")

plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "06_histogram_kde.png",
    dpi=300
)

plt.show()



# CHART 7 - BOX PLOT

plt.figure(figsize=(10, 5))

sns.boxplot(
    x=price_df
)

plt.title(
    "Box Plot of Book Prices"
)

plt.xlabel("Price (GBP)")

plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "07_boxplot.png",
    dpi=300
)

plt.show()



# CHART 8 - BAR CHART

publisher_counts = (
    df["publisher"]
    .value_counts()
    .head(10)
    .sort_values()
)

plt.figure(figsize=(11, 6))

publisher_counts.plot(
    kind="barh"
)

plt.title(
    "Top 10 Publishers by Number of Books"
)

plt.xlabel("Number of Books")
plt.ylabel("Publisher")

plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "08_bar_publishers.png",
    dpi=300
)

plt.show()



# CHART 9 - PIE CHART

type_counts = (
    df["type"]
    .value_counts()
    .head(5)
)

plt.figure(figsize=(8, 8))

plt.pie(
    type_counts.values,
    labels=type_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title(
    "Distribution of Top 5 Book Types"
)

plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "09_pie_types.png",
    dpi=300
)

plt.show()


print("\n" + "=" * 70)
print("DATA VISUALIZATION COMPLETED")
print("=" * 70)
```
